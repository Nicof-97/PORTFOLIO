```python
# Ver todos los valores > 1.0
outliers = df[df['RevolvingUtilizationOfUnsecuredLines'] > 1.0]['RevolvingUtilizationOfUnsecuredLines']

```

```python
# Ver cuántos están en cada rango
print("Valores entre 1.0 y 1.1:")
print(len(outliers[outliers <= 1.1]))

print("\nValores entre 1.1 y 2.0:")
print(len(outliers[(outliers > 1.1) & (outliers <= 2.0)]))

print("\nValores entre 2.0 y 10.0:")
print(len(outliers[(outliers > 2.0) & (outliers <= 10.0)]))

print("\nValores > 10.0:")
print(len(outliers[outliers > 10.0]))

# Ver algunos ejemplos de cada rango
print("\n--- Ejemplos entre 1.0 y 1.1 ---")
print(outliers[outliers <= 1.1].head(10).values)

print("\n--- Ejemplos entre 1.1 y 2.0 ---")
print(outliers[(outliers > 1.1) & (outliers <= 2.0)].head(10).values)

print("\n--- Ejemplos > 10.0 ---")
print(outliers[outliers > 10.0].head(10).values)

```

Aca se entiende que los valores entre 1.0 y 1.1 son valores que estan en el limite y no se reemplazan
Por otra parte los valores entre 1.1 y 10.0 son posibles errores de escala por lo que los dividimos por 100
Por último los valores mayores 10.0 son errores graves que seran reemplazados por 1.0

```python
# Valores entre 1.1 y 10.0: Posibles errores de escala → dividir por 100
df.loc[(df['RevolvingUtilizationOfUnsecuredLines'] > 1.1) & 
       (df['RevolvingUtilizationOfUnsecuredLines'] <= 10.0), 
       'RevolvingUtilizationOfUnsecuredLines'] = df['RevolvingUtilizationOfUnsecuredLines'] / 100

# Valores > 10.0: Errores graves → reemplazar con 1.0
df.loc[df['RevolvingUtilizationOfUnsecuredLines'] > 10.0, 
       'RevolvingUtilizationOfUnsecuredLines'] = 1.0
```

Limpiando valores de la columna DebtRatio

```python
print("Valores > 5.0:")
print(len(df[df['DebtRatio'] > 5.0]))

print("\nValores > 10.0:")
print(len(df[df['DebtRatio'] > 10.0]))

print("\nValor máximo:")
print(df['DebtRatio'].max())

print("\nPercentil 99:")
print(df['DebtRatio'].quantile(0.99))
```

Observamos que el número de valores por encima de 5.0 y de 10.0 es muy alto y que el valor máximo tambien por lo que veremos si el problema esta en esos 5% de percentil o más

```python
print("Percentil 95:")
print(df['DebtRatio'].quantile(0.95))
```

En este caso habiamos observado que no eran solo el 5% por lo que podemos deducir que uno de los problemas por el cual el debratio daba tan alto es que el monthlyincome era de 0 entonces eso potenciaba todos los valores, por lo que decidimos eliminar el monthlyincome = 0 como lo vemos a continuación

```python
print("Valores de MonthlyIncome = 0:")
print(len(df[df['MonthlyIncome'] == 0]))

print("\nValores de MonthlyIncome entre 0 y 500:")
print(len(df[(df['MonthlyIncome'] > 0) & (df['MonthlyIncome'] < 500)]))
```

```python
# Paso 1: Eliminar MonthlyIncome = 0
df = df[df['MonthlyIncome'] > 0]
```
